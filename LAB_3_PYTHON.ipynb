```python
# a student's marks are given in a list. 
# the system needs to increase mark by 5 bonus marks, subject to a maximum of 100 marks.
marks = list(map(int, input("Enter the marks of students separated by space: ").split()))
up_marks = list(map(lambda x: min(x + 5, 100), marks))
    
print(f"original marks are: {marks}")
print(f"The updated marks are: {up_marks}")
```

```python
# the univerity provides a scholarship to students based on their marks.
#  who score at least 80 marks are eligible for scholarship.
# should also take names and print only the names of students who are eligible for scholarship. using lambda function and filter function.
names = input("Enter the names of students separated by space: ").split()
marks = list(map(int, input("Enter the marks of students separated by space: ").split()))

scholarship_eligible = list(filter(lambda x: marks[names.index(x)] >= 80, names))

print(f"Original student records: {list(zip(names, marks))}")
print(f"Students eligible for scholarship: {scholarship_eligible}")

```

```python
# university has student records containg name and marks. 
# The system needs to sort the records based on marks in descending order.
students = [("vivek chinmay", 85), ("varsha", 78), ("chinmay", 92), ("vivek", 88)]
sorted_students = sorted(students, key=lambda x: x[1], reverse=True)    

print(f"Original student records: {students}")
print(f"Sorted students by marks: {sorted_students}")
```

```python
# an online store has a list of product prices.
# the system must :
# add 18% GST to each product price.
#consider only products costing more than 500 after adding GST.
# calculate the total using reduce function.
from functools import reduce    
prices = list(map(float, input("Enter the product prices separated by space: ").split()))
gst_prices = list(map(lambda x: x + (x * 0.18), prices)) # adding 18% GST to each product price 
filtered_prices = list(filter(lambda x: x > 500, gst_prices)) # filtering products costing more than 500 after adding GST
total = reduce(lambda x, y: x + y, filtered_prices,0) # calculating the total using reduce function
print(f"Original product prices: {prices}")
print(f"GST prices: {gst_prices}")
print(f"Filtered prices: {filtered_prices}")
print(f"Total cost of products after GST and filtering: {total}")
# reduce function is used to apply a rolling computation to sequential pairs of values in a list. 
# It takes two arguments: a function and an iterable. The function is applied cumulatively to the items 
# of the iterable, from left to right, so as to reduce the iterable to a single value. 
# In this case, we are using it to calculate the total cost of products after adding GST and 
# filtering out those that cost more than 500.
```

```python
''' Scenario 2: student performance analysis- *args
an mca student may study a diffrent number of subjects depending on the semester.
the university wants a program that can calculate the students percentage and grade regardless of the number of subjects.

problem statement: develop a function using *args that:
1. accepts marks for any number of subjects.
2. calculates the total and percentage.
3. determines the grade based on the percentage.
4. returns the total, percentage and grade.
'''
```

```python
def student_performance(name, *marks):
    total = sum(marks)
    percentage = total / len(marks)

    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    return total, percentage, grade


name = input("Enter student name: ")

marks = []

while True:
    subject = input("Enter subject name (or 'done' to finish): ")

    if subject.lower() == "done":
        break

    mark = float(input("Enter marks for " + subject + ": "))
    marks.append(mark)


total, percentage, grade = student_performance(name, *marks)

print("\n--- Student Result ---")
print("Name:", name)
print("Total:", total)
print("Percentage:", percentage)
print("Grade:", grade)
```

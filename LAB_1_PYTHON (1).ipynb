### CASE 1 SHOPPING CART BILL

```python
print("------------------------------")
print("\tSHOPPING CART")
print("------------------------------")
nam=input("CUSTOMER NAME:")
pro_nam=input("PRODUCT NAMR:")
qua=int(input("QUANTITY:"))
price=float(input("PRICE:"))
gst=float(input("GST (%):"))
```

```python
gst_div=gst/100
total_cost=qua*price
gst_am=total_cost*gst_div
final_am=gst_am+total_cost
```

```python
print("------------------------------")
print("\tSHOPPING BILL")
print("------------------------------")
print("Customer name:",nam)
print("Product name:",pro_nam)
print("Quantity:",qua)
print("Price/item:",price)
print("Total cost:",total_cost)
print("GST amount:",gst_am)
print("Final amount:",final_am)
```

### CASE_03 loops

```python
# using range function

a=[4,2,1,0,9]
key=int(input("Enter the key:"))

for i in range(len(a)):
    if a[i] == key:
        pos=i+1
        print("Found at position",pos)
        break
else:
    print("Not Found")
```

```python
# using in 

a=[4,2,1,8,9]
key=int(input("Enter the key:"))

for i in a:
    if i == key:
        print("Found")
        break
else:
    print("Not Found")
```

```python
str="vivek"
for i in str:
    print(i)
```

```python
# enumerate function

a=["red","blue","green"]

for index,i in enumerate(a):
        print(index,i)
```

```python
# zip function to merge multivalued lists

a=["red","blue","green"]
b=["100","200","300"]

for name,value in zip(a,b):
        print(name,value)
```

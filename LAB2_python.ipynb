```python
n = int(input("Enter number of rows: "))

for i in range(n):
    num = 1
    print(" " * (n - i), end="")

    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)

    print()
```

```python
# combining break and continue

n=[11,2,3,4,5]

for n in n:
    if n%2!=0:
        continue
    print("first even number:",n)
    break
    

        
```

```python
while True:
    
    print("1. Addition\n2. Subtraction\n3. Multiplication\n4. Exit\n")
    ch = int(input("Enter your choice: "))
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    if ch == 1:
        print(f"Sum of {a} + {b} is {a + b}")
    elif ch == 2:
        print(f"Subtraction of {a} - {b} is {a - b}")
    elif ch == 3:
        print(f"Multiplication of {a} * {b} is {a * b}")
    elif ch == 4:
        print("exiting!")
        break  
    else:
        print("Invalid choice! Please try again.")
```
